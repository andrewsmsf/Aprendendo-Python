# Arithmetric Operators

```python
# Sao usados para performar operacos em valores

a = 2 * 5
b = 3 + 7
c = 16 - 9
d = 40 / 4

print(f'a  = {a}')
print(f'b  = {b}')
print(f'c  = {c}')
print(f'd  = {d}')
```

```python
# result = dividend % divisor  dividend numer being divided by divisor
# divisor, the number by witch the dividend is divided.
# result, the remainder of the dividsion.

a = 32
b = 7
c = 55

d = a % 8
e = b % 3 
f = c % 11

print(f'a = {a}')
print(f'b = {b}')
print(f'c = {c}')
print(f'd = {d} its a even number')
print(f'e = {e} its a odd number')
print(f'f = {f} its a even number')
```

```python
"""
The modulo operator is "%"
The remainder is the amount left over after division. 

If the number is divisible by 2 and the remainder is 0, then the number is even.
If the number is divisible by 2 and the remainder is 1, then the number is odd.

to check the code, we use the == operator, that checks if two values are equal.
Ex.: result == 0 ask "Is result equal to 0?"

"""

a = 10 + 2
b = 23 - 7
result = a % b # 12 % 16

# Para verificar se eh par, dividimos por 2 e verificamos se o resto eh 0
is_even = result % 2 == 0 # 12 % 2 = 0. 0 eh igual ao 0, ent o resultado eh True

print(f'a = {a}')
print(f'b = {b}')
print(f'result = {result}') # the result is 12 becouse 16 cant fit in 12, so the remainder is 12
print(f'is_even = {is_even}')
```
