```python
# 3. Orçamento apertado — Crie salario e gasto_total (use valores realistas). Verifique se a pessoa não estourou o orçamento (gasto <= salario).
salario = None
aluguel = 1210.55
comida = 532.99
luz = 320.20
internet = 175.50

print("Bem-vindo a sua calculadora de orcamento.\n")
print(f'Seu aluguel: R${aluguel}')
print(f'Gastos com comida: R${comida}')
print(f'Gastos com luz: R${luz}')
print(f'Gastos com internet: R${internet}')
salario = float(input("Qual seu salario? R$"))

gasto_total = aluguel + comida + luz + internet
lucro = salario - gasto_total

print(f'\nSeu salario = R${salario}')
print(f'Seu gasto total: R${gasto_total}')

if (gasto_total <= salario): 
    print("\nParabens voce nao estorou esse mes, lucro de R$", lucro)
    print(f'Quantas vezes o gasto cabe dentro do seu salario? {salario / gasto_total}')
else:
    print("\nVoce estorou o orcamento. Falta: R$", gasto_total - salario)



```

```python
# 4. Desconto ou não? — Crie valor_compra e limite_desconto. Verifique se o valor da compra ultrapassa o limite para ganhar desconto.
valor_compra = float(input("Qual valor de compra do produto que escolheu? R$"))

limite_desconto = 150.0

if (valor_compra <= limite_desconto):
    print(f'Parabens vc esta apto para a compra pois o valor da compra nao excedeu o limite {limite_desconto}')
    print(f'O preco da compra foi R$ {valor_compra}')
else:
    print(f'Voce excedeu o limie, o limite eh R${limite_desconto} e o preco da compra foi R${valor_compra}')
```

```python
#5. Senha correta? — Crie senha_cadastrada e senha_digitada (textos, use aspas). Verifique se diferem — ou seja, se a senha está errada.
print("Bem vindo ao sistema de senha. Objetivo apenas fazer a verificacao da senha 2 vezes.\nUse caracteres 'Hs_~!#.,' ")
senha_digitada = str(input("Digite sua senha: "))
senha_digitada2 = str(input("Digite sua senha novamente: "))

senha_cadastrada = senha_digitada == senha_digitada2

if (senha_cadastrada == True):
    print(f'Sua senha esta correta!')
else:
    print(f'Sua senha esta errada.')
```

```python
# correcao da 5

"""
Como fazer o != (a parte que você não sabia)
A lógica é o oposto do ==. No seu exercício 5, em vez de guardar senha_digitada == senha_digitada2 numa variável e comparar de novo, seria:
senha_cadastrada = "abc123"
senha_digitada = "abc123"

senha_errada = senha_digitada != senha_cadastrada   # False, senha bateu
O != responde True quando os valores são diferentes. Se as senhas fossem diferentes, daria True — "está errada". Simples assim.

"""
print("Bem vindo ao sistema de senha. Objetivo apenas fazer a verificacao da senha 2 vezes.\nUse caracteres 'Hs_~!#.,' ")
senha_cadastrada = input("Digite sua senha: ")
senha_digitada = input("Digite sua senha novamente: ")

senha_errada = senha_cadastrada != senha_digitada # o '!=" daria True quando os valores fosse diferentes

print(f'Sua senha esta errada? {senha_errada}')
```

```python
# 6. Percentual de meta — Crie meta_mensal e valor_alcancado. Verifique se o valor alcançado é maior ou igual à meta, mas também verifique se é estritamente menor que a meta. Use duas comparações separadas.
meta_mensal = int(input("Quantas metas voce quer tentar bater este mes? "))
valor_alcancado = int(input("Quantas voce conseguiu neste mes? "))


if(valor_alcancado >= meta_mensal):
    print(f'Parabens voce conseguiu acancar a meta de {valor_alcancado}/{meta_mensal}')
else:
    print(f'Voce esta abaixo da meta {valor_alcancado}/{meta_mensal}')
```

```python
# correcao da 6

meta_mensal = int(input("Quantas metas voce quer tentar bater este mes? "))
valor_alcancado = int(input("Quantas voce conseguiu neste mes? "))


if valor_alcancado >= meta_mensal:
    print(f'Parabens voce conseguiu acancar a meta de {valor_alcancado}/{meta_mensal}')
else:
    abaixo = valor_alcancado < meta_mensal
    print(f'Voce esta abaixo da meta {abaixo}')
```
